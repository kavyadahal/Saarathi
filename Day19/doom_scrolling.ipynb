```python
import pandas as pd
from matplotlib import pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('sleep_doomscrolling_habits.csv')
```

```python
df.head()
x = (np.array(df['caffeine_intake_mg_per_day']))
y = (np.array(df['doomscroll_sessions_per_night']))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.regplot(x=x, y=y, line_kws={'color': 'red'})
plt.xlabel('Caffeine intake (mg/day)')
plt.ylabel('Doomscroll sessions per night')

slope, intercept, r, p_value, std_err = stats.linregress(x, y)

if slope > 0:
    direction = "increases"
else:
    direction = "decreases"

plt.title(f'As caffeine goes up, doomscrolling {direction} (slope = {slope:.3f})')
plt.show()
```
