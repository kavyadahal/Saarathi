# Day 13 Revision Summary — Your First Models (Linear & Logistic Regression)

- Every scikit-learn model shares the same API — `model.fit(X_train, y_train)` to learn, `model.predict(X_test)` to guess. Learn it once, reuse it forever.
- **Linear regression** predicts a number. Score it honestly on the test set with **RMSE** (typical error, same units as `y`), **MAE** (average absolute miss), and **R²** (fraction of variance explained; 0 = no better than always guessing the mean) — always compare to a "predict the mean" baseline.
- **Logistic regression** predicts a category via a probability that gets rounded to a class. Read its results with a **confusion matrix**: `[[TN, FP], [FN, TP]]`.
- **Precision** (`TP/(TP+FP)`, punishes false alarms), **recall** (`TP/(TP+FN)`, punishes misses), and **F1** (harmonic mean of the two) tell a fuller story than a single accuracy number.
- **Accuracy lies on imbalanced data** — always predicting the majority class can score 98% accuracy with 0% recall. Never report accuracy alone; decide whether to optimise for precision or recall based on the real-world cost of a false alarm vs. a miss (e.g. cancer/fraud → maximise recall, spam filter → protect precision).
