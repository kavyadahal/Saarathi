# Day 3 Revision Summary — Loops: for, while, break/continue & Accumulate

- `for` loops walk through a sequence one item at a time; the loop variable holds the current item on each pass.
- `range(n)` / `range(a, b)` / `range(a, b, step)` generate number sequences — the stop value is always **excluded** (classic off-by-one trap: `range(5)` gives 0,1,2,3,4).
- `while` loops repeat until a condition becomes `False` — you must change the condition inside the loop or it runs forever (`Ctrl+C` escapes a runaway loop).
- `break` exits a loop immediately; `continue` skips straight to the next iteration.
- The **accumulate pattern** — start with an empty total, update it each pass, use the result after the loop — underlies sums, counts, and building up strings/lists. This is the single most important idea of the day.
