# Day 9 Revision Summary — OOP + numpy & pandas

- **Classes bundle data + behaviour**: `__init__` sets up attributes, `self` is the specific object (always the first parameter — forgetting it is the #1 class bug), and methods are just functions that live on the object and can use everything you already know about functions (hints, `return`, defaults).
- **Class vs dict judgement**: use a plain **dict** for data only (config, a JSON response, a row); use a **class** when data comes bundled with behaviour. A class with only `__init__` and attributes is "a dict wearing a costume"; a dict passed to six functions wants to be a class.
- **Inheritance, kept to one level**: a child class picks up the parent's attributes/methods, and `super().__init__(...)` calls the parent's setup — you already used this pattern with `class MyError(Exception)` on Day 7.
- **numpy arrays** do maths on the whole array at once — no loop needed (`np.array([1,2,3]) * 2` doubles every element, unlike `[1,2,3] * 2` which just repeats the list); `.shape`, `.mean()`, `.sum()`, `.max()`, and boolean filtering (`a[a > 20]`) are the core toolkit.
- **pandas DataFrames** are built from a dict of lists; `.head()`, column selection, boolean-mask row filtering, `df.describe()`, and `to_csv()`/`read_csv()` round out the data toolkit that all of Week 3 and classical ML builds on.
