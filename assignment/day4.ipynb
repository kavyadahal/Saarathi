# Day 4 Revision Summary — Collections: Lists, Dictionaries & Sets

- Lists hold ordered collections; built-ins `sum()`, `min()`, `max()`, `len()` operate directly on them, and `.append()` grows a list in place.
- Dictionaries store key -> value pairs; `d[key]` reads/writes a value, `d.items()` iterates pairs, and `d.get(key, default)` does a *safe* lookup that won't crash on a missing key.
- Sets automatically de-duplicate values — `set(some_list)` is the standard way to get unique values and `len(...)` on that set gives the count of distinct values.
- The Day 3 loop + accumulate pattern reappears here: building a word-frequency dictionary with `counts.get(word, 0) + 1`.
- Predicting dict/set output *before* running the code builds intuition for how Python handles duplicate keys, membership tests (`in`), and set de-duplication.
