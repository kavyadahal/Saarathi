# Day 1 Revision Summary — Variables, Types & Print Basics

- Variables store values under names (`str`, `int`, `float`, `bool`) and can be reassigned at any time.
- `print()` accepts multiple comma-separated arguments and automatically separates them with spaces.
- Comparison operators (`>`, `<`, `==`, etc.) evaluate to a boolean (`True`/`False`).
- `type()` reveals the underlying Python type of any variable or expression.
- Basic arithmetic (`*`, `/`, `-`, `+`) works on `int`/`float` values, e.g. the Celsius-to-Fahrenheit formula `F = C * 9/5 + 32`.
- f-strings are coming on Day 2 — for today, comma-separated `print()` calls are the tool of choice.
