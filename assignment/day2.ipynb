# Day 2 Revision Summary — Strings & Conditionals

- String methods: `.strip()` removes leading/trailing whitespace, `.title()` capitalizes each word, `.split()` breaks a string into a list of words, `len()` gives a length.
- f-strings (`f"...{var}..."`) are the clean, readable way to build formatted output.
- `if` / `elif` / `else` chains are evaluated top to bottom — the first `True` branch runs and the rest are skipped.
- Guard clauses: validate and reject bad input (e.g. an out-of-range score) *before* the "normal" branches run.
- Chained comparisons plus boolean logic (`and`) combine multiple conditions, e.g. applying a discount only when a flag AND a value are both true.
