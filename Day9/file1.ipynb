```python
import pandas as pd
import numpy as np
```

```python
#make an array of 6 scores. Print mean, max, and only those above the mean.
a = np.array([10,40,30,50,80,90])
print(f"Mean:{a.mean()},Max:{a.max()}")
b = a.mean()
for val in a:
    if val>b:
        print(val)
        
      
```

```python
import pandas as pd

data = pd.DataFrame({
    "name": ["Kavya", "Prachi", "Insaf", "Maheshwor"],
    "score": [80, 50, 90, 100],
})

print(data)
#Print .head(), the mean score, and rows where score > 70.
print(data.head())
print("Mean:",data["score"].mean())
for val in data["score"]:
    if val > 70:
        print(val)
#Add a passed column that's True when score ≥ 40.
data['passed'] = data['score'] > 40
print(data)
        
```
