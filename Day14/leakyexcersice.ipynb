```python
import numpy as np
from sklearn.model_selection import cross_val_score
from sklearn.feature_selection import SelectKBest , f_classif
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression


rng = np.random.RandomState(42) #random number generator function
X = rng.randn(200,1000) #200 rows #1000 columns
y = rng.randint(0,2,200) #upper limit (2) is excluded.
#y = rng.randint(low,high,size)
```

```python
X_sel = SelectKBest(f_classif, k=20).fit_transform(X, y)
leaky = cross_val_score(LogisticRegression(max_iter=1000), X_sel, y, cv=5)
```

```python
pipe = Pipeline([
    ("select", SelectKBest(f_classif, k=20)),
    ("clf", LogisticRegression(max_iter=1000)),
])
honest = cross_val_score(pipe, X, y, cv=5)
```

```python
print(leaky.mean(),honest.mean()) 
```
