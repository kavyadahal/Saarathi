# Digits : SGD VS ADAM


```python
import torch
import torch.nn as nn
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Load data
X, y = load_digits(return_X_y=True)

# Train/test split
X_tre, X_tes, y_tre, y_tes = train_test_split(
    X, y,
    train_size=0.8,
    random_state=42,
    stratify=y
)

# Standardization
sc = StandardScaler()
Xtr = torch.tensor(sc.fit_transform(X_tre), dtype=torch.float32)
Xtes = torch.tensor(sc.transform(X_tes), dtype=torch.float32)

# Labels
ytr = torch.tensor(y_tre, dtype=torch.long)
ytes = torch.tensor(y_tes, dtype=torch.long)

# Model
model = nn.Sequential(
    nn.Linear(64, 32),
    nn.ReLU(),
    nn.Linear(32, 10)
)

# Loss and optimizer
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.1)

# Training
for epoch in range(60):
    pred = model(Xtr)
    loss = loss_fn(pred, ytr)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if (epoch + 1) % 5 == 0:
        lr = optimizer.param_groups[0]['lr']
        print(f"Epoch {epoch+1:3d} | lr: {lr:.4f} | loss: {loss.item():.4f}")

    

# Evaluation
with torch.no_grad():
    pred = model(Xtes)
    acc = (pred.argmax(1) == ytes).float().mean()
    
    print("Accuracy:", acc.item())
    
```
