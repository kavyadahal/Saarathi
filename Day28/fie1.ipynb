# Auto Grad

```python
x = torch.tensor(3.0 , requires_grad=True)
y = x ** 2
y.backward()
print(x.grad)
```

```python
x_in  = torch.tensor(2.0)
target = torch.tensor(3.0)
w =  torch.tensor(0.5  , requires_grad=True)
b = torch.tensor(0.0 , requires_grad=True)

z = w * x_in + b
loss = (z-target) ** 2
loss.backward()
print(z.item() , loss.item())
print(w.grad , b.grad)
```
