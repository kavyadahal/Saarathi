```python
import pandas as pd
import numpy as np
```

```python
#Merge students.csv and scores.csv on sudent)id
df1 = pd.read_csv('students.csv')
df2 = pd.read_csv('scores.csv')
df = pd.merge(df1,df2,on = "student_id")
df
```

```python
#Average score per subject with goupby
df.groupby("subject")['score'].mean()
```

```python
#Average score per city(needs the merge first)
df.groupby("city")['score'].mean()
```

```python
#Filter L Which student scored >90 in any subject

df[df['score']>90]
```

```python
df = df.sort_values(by ='city',ascending = True)
len(df)
```

```python
df.to_csv('city_group.csv',index = False)
```
