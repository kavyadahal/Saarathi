"""
               print(df.groupby("Data_value").get_group(83802.180055))

In Python, decimals often have tiny hidden precision differences in memory (e.g., 83802.180055000003).
get_group() requires an exact binary match, so tiny floating-point variations cause a KeyError.
sO USE NUMPY to get to the closest near value
"""

```python
df= df.copy()
grouped = df.groupby("Data_value")
target_value = 83802.180055
exact_key = min(grouped.groups.keys(),key = lambda k : abs(float(k)-target_value))
result = grouped.get_group(exact_key)
print(result)
```
