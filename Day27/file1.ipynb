```python
import json
from openai import OpenAI

client = OpenAI(
    api_key="GROQ_API_KEY_REMOVED",  # console.groq.com/keys
    base_url="https://api.groq.com/openai/v1"
)

user_query = "The delivery was late but the product quality is excellent."

system_prompt = """
You are an expert review analyzer.

Analyze the user's review and respond with a JSON object in exactly this format:
{
    "review": "POSITIVE" or "NEGATIVE" or "NEUTRAL",
    "users_review": "<the user's original review>",
    "message": "<a polite, polished response to the user that also tells them to check out Sarathi's course>"
}

Use "NEUTRAL" when the review has a genuine mix of positive and negative points that don't clearly favor one side.
Return only the raw JSON object.
"""

response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_query}
    ],
    response_format={"type": "json_object"}
)

ai_response = response.choices[0].message.content
serialized_value = json.loads(ai_response)

print(json.dumps(serialized_value, indent=2))
```
